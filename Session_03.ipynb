# Loops for Strings

```python
# using index values

a = 'SHERYIANS'

for i in range(0, 9, 1):
  print(a[i])
```

```python
a = "SHERYIANS TEACHES INDUSTRY THINGS"

print(len(a))

for i in range(len(a)):
  print(a[i])
```

```python
# directly over the strings

a = "SHERYIANS IS COOL"

for i in a:
  print(i)
```

# Break and Continue

```python
# break

for i in range(1, 21):
  if i == 15:
    break
  else:
    print(i)
```

```python
# continue

for i in range(1, 21):
  if i == 15:
    continue
  else:
    print(i)
```

# Questions:

```python
# Accept an integer and print hello word n times.

num = int(input("Enter any integer: "))

for i in range(num):
  print("hello world")
```

```python
# print natural number up to n.
num = int(input("Enter any integer: "))
for i in range(1, num+1):
  print(i)
```

```python
# reverse for loop. Print n to 1.
num = int(input("Enter any integer: "))
for i in range(num, 0, -1):
  print(i)
```

```python
# take a number as input and print its table.

num = int(input("Enter any number: "))

for i in range(1, 11):
  print(f"{num} * {i} = {num*i}")
```

```python
# sum up to n terms.

a = 10
a = a+5

print(a)
```

```python
sum = 0
sum = sum + 1
sum = sum + 2
```

```python
n = int(input("Enter any number : "))
sum = 0
for i in range(1, n+1):
  sum = sum + i

print(f"your sum is {sum}")
```

```python
# factorial of a number.

n = int(input("Enter any number : "))
fact = 1
for i in range(1, n+1):
  fact = fact * i

print(f"your factorial is {fact}")
```

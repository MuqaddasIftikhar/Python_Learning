Question 1: Smart Shopping Budget


Concepts: Variables, arithmetic operators, conditions.

Imagine you're shopping with a budget of Rs. 5,000.

You buy three products:
- Headphones: Rs. 1,850
- Backpack: Rs. 1,250
- Notebook set: Rs. 450
Your task:
1. Calculate the total cost.
2. Calculate how much money remains.
3. If the total exceeds Rs. 4,000, apply a 10% discount.
4. Display the final bill and remaining budget.

Challenge: If the discount applies, recalculate the remaining budget using the discounted bill.

```python
budget = 5000

headphones = 1850
backpack = 1250
notebook_set = 450

# total_cost
total_cost = headphones + backpack + notebook_set
print(f"Total Cost is Rs.{total_cost}")

# remaining amount
remaining_amount = budget - total_cost
print(f"Remaining Amount is Rs.{remaining_amount}")

# discount
discount = 0

if total_cost > 4000:
  discount = total_cost * (10/100)
  print(f"Your discount is Rs.{discount}")
else:
  print("No discount! Your budget is in the range of Rs.4000")

# Final Result
final_bill = total_cost - discount
remaining_amount = budget - final_bill
print(f"Your final bill is Rs.{final_bill} and your remaining budget is Rs.{remaining_amount}")
```

Question 2: The Three-Attempt Password System

Concepts: while loop, conditional statements, comparison operators.

Imagine you're creating a simple login system.

The correct password is "python123".

Your task:
1. Give the user a maximum of three attempts to enter the password.
2. If the password is correct, display "Login successful!".
3. If the password is incorrect, display "Incorrect password!".
4. If all three attempts fail, display "Account locked!".
5. Once the correct password is entered, don't ask for the password again.

Challenge: Display the number of attempts remaining after each incorrect entry.

Hint: Think about how a while loop can stop when either the password is correct or the attempts run out.

```python
password = "python123"

attempts = 0

login_successful = False

while attempts < 3:
  user_input = input("Enter your password: ")
  if password == user_input:
    print("Login successful!")
    login_successful = True
    break
  else:
      print("Login Failed!")
      attempts = attempts + 1

if login_successful == False:
      print("Account Locked!")

```

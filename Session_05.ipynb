# Functions

```python
# built in functions

print("Hello, How are you?")
```

```python
# user define functions

def hello():
  print("This is a hello function so I am doing hello")

hello()
```

```python
# Function parameters and arguments

# positional arguments
def sum(a, b):
  print(f"The sum of your numbers is: {a + b}")

print(sum(12, 12))
print(sum(45, 45))
```

```python
# keyword arguments

def hello(name, age):
  print(f"Your name is {name} and your age is {age}")

hello(age = 22, name = "Muqaddas")
```

```python
# default arguments

def sum(a, b=45):
  print(f"The sum is {a+b}")
sum(12)
sum(12, 34)
```

```python
# check if the string is palindrome or not using functions

def pallindrome(st):
  rev = 0
  for i in range(len(st)-1, -1, -1):
    rev = rev + st[i]

  if rev == st:
    print("Pallindrome")
  else:
    print("Not a pallindrome")

pallindrome("Naman")
pallindrome("Cursor")
```
